# Bench Marking Models on 3 Custom Datasets
## Info on Dataset
### Custom Dataset 1
* Train - 8378 images
* Test - 65 images
* Val - 1505 images
### Custom Dataset 2
* Train - 10359 images
* Test - 1470 images
* Val - 2922 images


```python
MODEL_PATH_1 = r"D:\training_nano_model\nano_models\1_nano.pt"
MODEL_PATH_2 = r"D:\training_nano_model\nano_models\2_nano.pt"
```

```python
DATA_PATH_1 = r"D:\downloads\merged.v2i.yolov11\data.yaml"
DATA_PATH_2 = r"D:\second_run_on_yolo\yolo_train_logs_2\My First Project.v3i.yolov11\data.yaml"
```

```python
import ultralytics
ultralytics.__version__
```

```python
from ultralytics import YOLO
import torch
import yaml
import matplotlib.pyplot as plt
from pathlib import Path

print("CUDA available:", torch.cuda.is_available())
print("CUDA device:", torch.cuda.current_device())
print("CUDA device name:", torch.cuda.get_device_name(torch.cuda.current_device()))
```

```python
from ultralytics.data.utils import check_det_dataset

check_det_dataset(DATA_PATH_1)
check_det_dataset(DATA_PATH_2)
```

```python
# TEST SET METRICS(mAP@0.5 & mAP@0.5:0.95)OF DIFFERENT MODELS ON CUSTOM DATASET 1
model_1 = YOLO(MODEL_PATH_1)
model_2 = YOLO(MODEL_PATH_2)


model_1_metrics = model_1.val(data=DATA_PATH_1, split='test')
model_2_metrics = model_2.val(data=DATA_PATH_1, split='test')




print("\nTest Set Metrics of 2 models on Custom Dataset 1:")
print("mAP@0.5 of 1_nano.pt:", model_1_metrics.box.map50)
print("mAP@0.5:0.95 of 1_nano:", model_1_metrics.box.map)


print("mAP@0.5 of 2_nano.pt:", model_2_metrics.box.map50)
print("mAP@0.5:0.95 of 2_nano.pt:", model_2_metrics.box.map)


```

```python
# TEST SET METRICS(mAP@0.5 & mAP@0.5:0.95)OF DIFFERENT MODELS ON CUSTOM DATASET 2
model_1 = YOLO(MODEL_PATH_1)
model_2 = YOLO(MODEL_PATH_2)

model_1_metrics = model_1.val(data=DATA_PATH_2, split='test')
model_2_metrics = model_2.val(data=DATA_PATH_2, split='test')




print("\nTest Set Metrics of 2 models on Custom Dataset 2:")
print("mAP@0.5 of 1_nano.pt:", model_1_metrics.box.map50)
print("mAP@0.5:0.95 of 1_nano:", model_1_metrics.box.map)


print("mAP@0.5 of 2_nano.pt:", model_2_metrics.box.map50)
print("mAP@0.5:0.95 of 2_nano.pt:", model_2_metrics.box.map)


```
