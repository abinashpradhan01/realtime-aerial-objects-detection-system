```python
import pandas as pd
from torch.utils.tensorboard import SummaryWriter

# Path to results.csv
csv_path = "D:/retraining_yolo/yolo_train_logs_2/yolov11m_exp2_colab/results.csv"
log_dir = "D:/retraining_yolo/yolo_train_logs_2/yolov11m_exp2_colab/tensorboard_post"

# Load CSV
df = pd.read_csv(csv_path)
writer = SummaryWriter(log_dir=log_dir)

# Write scalar logs to TensorBoard
for epoch, row in df.iterrows():
    writer.add_scalar("Loss/Box", row["      box_loss"], epoch)
    writer.add_scalar("Loss/Class", row["     cls_loss"], epoch)
    writer.add_scalar("Loss/DFL", row["      dfl_loss"], epoch)
    writer.add_scalar("Metrics/mAP50", row["metrics/mAP50(B)"], epoch)
    writer.add_scalar("Metrics/Precision", row["         metrics/precision(B)"], epoch)
    writer.add_scalar("Metrics/Recall", row["         metrics/recall(B)"], epoch)

writer.close()
print("✅ TensorBoard logs written!")
```
