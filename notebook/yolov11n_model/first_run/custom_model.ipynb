## Training yolov11(nano)

```python
import ultralytics
ultralytics.__version__
```

```python
from ultralytics import YOLO

# Load a COCO-pretrained YOLO11n model
model = YOLO("yolo11n.pt")
```

```python
data_yaml_path = r"D:\downloads\merged.v2i.yolov11\data.yaml"
```

```python
from ultralytics.data.utils import check_det_dataset

check_det_dataset(data_yaml_path)
```

```python
results = model.train(
    data=data_yaml_path,
    epochs=100,
    batch=5,
    name="yolov11n_1",
    project=r"yolo_nano_logs_1",
    device='cuda:0',  
    save=True,
    patience=10,
    verbose=True,
    workers=2,
    resume=False,
    exist_ok=True,
    imgsz=640,        # Standard resolution for detection, balanced performance/speed
    lr0=0.01,         # Initial learning rate – works well with SGD
    momentum=0.937,   # Helps stabilize gradient updates
    weight_decay=0.0005  # Prevents overfitting, helps regularize model

)
```

```python
metrics = model.val(data=data_yaml_path, split='test' )  # evaluating model performance on the validation set
```

```python
# printing metrics of our model
print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
```

```python
# predicting on an image
results = model.predict(source="sample.jpg", save=True, conf=0.25)
```
